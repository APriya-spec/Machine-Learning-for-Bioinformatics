```python
import numpy as np
from sklearn.decomposition import PCA

# Step 1: Define the 2D dataset
X = np.array([
    [0, 2],
    [2, 4],
    [4, 4],
    [5, 4],
    [7, 6]
])

# Step 2: Apply PCA
pca = PCA(n_components=2)
pca.fit(X)

# Step 3: Print the weight vectors (principal components)
print("Principal Components (weight vectors):")
print(pca.components_)

# Optional: Explained variance ratio
print("\nExplained variance ratio:")
print(pca.explained_variance_ratio_)
```
